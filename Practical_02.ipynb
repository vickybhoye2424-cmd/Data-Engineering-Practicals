## Aim
To perform the ETL process in Power BI by extracting, transforming, and loading data to prepare it for analysis.

## Step 1 — Import Excel Data into Power BI
The Excel workbook containing the Products, Orders, and Order_Details tables was imported into Power BI for further data transformation and analysis.

## Step 2 — Open Power Query Editor
Power Query Editor was opened to perform the required data transformation operations on the imported tables.

## Step 3 — Remove Unnecessary Columns
Unnecessary columns were removed from the Products table to retain only the fields required for further analysis.

## Step 4 — Verify the Data Type of UnitsInStock
The data type of the UnitsInStock column was verified as Whole Number, which is appropriate for representing stock quantities.

## Step 5 — Calculate Line Total
A new LineTotal column was created by multiplying UnitPrice by Quantity for each order detail.

**Formula:** `LineTotal = UnitPrice × Quantity`

## Step 6 — Rename and Reorder Columns
The columns were verified and arranged in a suitable order for further analysis.

## Step 7 — Load the Transformed Data
The transformed data was loaded into Power BI Desktop using the Close & Apply option.

## Step 8 — Create Relationships Between Tables
Relationships were created between the related tables using their common key fields to establish the data model.

## Step 9 — Result
The ETL process was successfully performed in Power BI by extracting data from an Excel workbook, transforming the data using Power Query, creating the Line Total column, loading the transformed data, and establishing relationships between the tables.
